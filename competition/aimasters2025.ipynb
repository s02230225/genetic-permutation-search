# Задача на оптимизацию

```python
# матрица для оптимизации
import pandas as pd
X = pd.read_csv('X_aim.csv', header=None, index_col=False).values
X.shape
```

```python
X[0][0]
```

```python
# функция для оптимизации
import numpy as np
def func4opt(I):
    global X
    assert len(set(I)) == 100
    z = X[:, I[0]] * len(I)
    for i in I[1:]:
        z += np.sin(10 * X[:, i] + z)
    return np.sum(z ** 2)
```

```python
# значение функции на списке [0,1,...,99]
I = list(range(100))
func4opt(I)
```

```python
# значение функции на случайном списке
np.random.shuffle(I)
func4opt(I)
```
